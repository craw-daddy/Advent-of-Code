## Part 1

```python
with open('data/08-Dec.input', 'r') as f:
    data = f.read().strip()
```

```python
data = [tuple(int(y) for y in x.split(',')) for x in data.split('\n')]

d_squared = []

for i,j in itertools.combinations(data, 2):
    d_squared.append((i,j,((i[0] - j[0])**2 + (i[1] - j[1])**2 + (i[2] - j[2])**2)))

d_squared = sorted([(x[2], x[0], x[1]) for x in d_squared])
```

```python
circuits = { v : k for k,v in enumerate(data) }
```

```python
limit = 1000

for i in range(limit):
    t_1, t_2 = d_squared[i][1], d_squared[i][2]
    values = [circuits[t_1], circuits[t_2]]
    new_key = min(values)
    
    for junction in circuits.keys():
        if circuits[junction] in values:
            circuits[junction] = new_key

c = sorted(list(set(circuits.values())))

groups = sorted([len([k for k,v in circuits.items() if v == z]) for z in c], reverse=True)

print(reduce(lambda x,y: x*y, groups[:3]))
```

```python
circuits = { v : k for k,v in enumerate(data) }

i = 0

while len(set(circuits.values())) > 1:
    t_1, t_2 = d_squared[i][1], d_squared[i][2]
    values = [circuits[t_1], circuits[t_2]]
    new_key = min(values)
    
    for junction in circuits.keys():
        if circuits[junction] in values:
            circuits[junction] = new_key

    i = i + 1

print(t_1[0]*t_2[0])
```
